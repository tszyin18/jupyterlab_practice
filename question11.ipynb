```python
def compute_mean(lst):
    return sum(lst) / len(lst)

print(compute_mean([2,4,6,8]))
```

The function take a list of numbers and returns the sum of the list of numbers, divide by the number of numbers in the list of numbers to calculate the mean of the list of numbers.

$$
\bar{x}=\frac{x_1+x_2+\cdots+x_n}{n} = \frac{1}{n} \Bigg( \sum^{n}_{i=1} x_i \Bigg)
$$
